## 📌 Project Overview
This notebook demonstrates a complete end-to-end Machine Learning workflow to predict housing prices in California. We use the classic **California Housing Dataset** from `scikit-learn` and train a **Random Forest Regressor** to capture non-linear relationships in the data.

### 📊 Key Workflow Steps:
1. **Data Ingestion & Exploration**
2. **Train-Test Split**
3. **Model Training (Random Forest)**
4. **Performance Evaluation (R² & MAE)**
5. **Feature Importance Breakdown**
6. **Real-time Inference & Serialization**
7.

```python
# =====================================================================
# PROJECT: CALIFORNIA HOUSING PRICE PREDICTION
# MODEL: RANDOM FOREST REGRESSOR (BASELINE)
# AUTHOR: ANIKET RAJ
# =====================================================================

# --- Step 1: Import Core Libraries ---
import pandas as pd
import numpy as np
from sklearn.datasets import fetch_california_housing

# --- Step 2: Load and Prepare the Dataset ---
housing = fetch_california_housing(as_frame=True)
df = housing.frame

# Inspect the first few rows of the dataset
print("--- Dataset Head ---")
print(df.head())

# Check data types and look for structural information
print("\n--- Dataset Info ---")
print(df.info())

# --- Step 3: Separate Features (X) and Target Variable (y) ---
x = df.drop(columns=['MedHouseVal'])
y = df['MedHouseVal']

print("\n--- Feature Columns ---")
print(x.columns.tolist())
print(f"Target Shape: {y.shape}")

# --- Step 4: Split Dataset into Training and Testing Sets ---
from sklearn.model_selection import train_test_split

# 80% Training and 20% Testing split for model validation
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print(f"\nTraining Set Shape: {X_train.shape}")
print(f"Testing Set Shape: {X_test.shape}")

# --- Step 5: Initialize and Train Random Forest Regressor ---
from sklearn.ensemble import RandomForestRegressor

# Initializing Random Forest with 100 decision trees
model_rf = RandomForestRegressor(n_estimators=100, random_state=42)

# Training the model on the training dataset
model_rf.fit(X_train, y_train)

# Generating predictions on the test dataset
y_pred = model_rf.predict(X_test)
print("\n--- Sample Predictions ---")
print(y_pred[:10])  # Printing the first 10 predicted values

# --- Step 6: Model Evaluation Metrics ---
from sklearn.metrics import mean_absolute_error, r2_score

r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

print("\n--- Model Performance Metrics ---")
print(f"R2 Score (Variance Explained): {r2 * 100:.2f}%")
print(f"Mean Absolute Error (MAE): {mae:.4f}")

# --- Step 7: Feature Importance Analysis (Statistical Insights) ---
# Extracting feature importances based on variance reduction
importances = model_rf.feature_importances_

# Constructing a structured DataFrame for clear scannability
feature_imp_df = pd.DataFrame({
    'Feature': x.columns,
    'Importance_Score': importances
}).sort_values(by='Importance_Score', ascending=False)

print("\n--- Feature Importance Breakdown ---")
print(feature_imp_df)
# --- Step 7.1: Visualize Feature Importances (Colorful Charts) ---
import matplotlib.pyplot as plt
import seaborn as sns

# Setting the aesthetic style for a clean, professional look
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Creating a colorful bar plot using the 'viridis' color palette
sns.barplot(
    x='Importance_Score', 
    y='Feature', 
    data=feature_imp_df, 
    palette='viridis',
    hue='Feature',
    legend=False
)

# Customizing the labels and title for maximum scannability
plt.title('Feature Importance Breakdown (Random Forest)', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Importance Score (Variance Reduction)', fontsize=12, fontweight='bold')
plt.ylabel('Dataset Features', fontsize=12, fontweight='bold')
plt.tight_layout()

# Displaying the colorful chart inside the notebook
plt.show()

# --- Step 8: Inference (Predicting Price for a New House) ---
# Input vector for a single house block
new_house_data = [[8.0, 25.0, 6.0, 1.0, 1200.0, 2.5, 37.88, -122.23]]

# Converting the list to a DataFrame with valid feature names to avoid warnings
new_house_df = pd.DataFrame(new_house_data, columns=x.columns)
predict_price = model_rf.predict(new_house_df)

# Displaying the final predicted value with proper scaling ($100k factor)
print("\n--- Real-Time Inference ---")
print(f"Predicted House Value for New Data: ${predict_price[0] * 100000:.2f}")

# --- Step 9: Model Serialization (Saving to Disk) ---
import joblib

# Exporting the trained model instance as a .pkl file for deployment
joblib.dump(model_rf, 'california_housing_rf_model.pkl')
print("\n[SUCCESS] Model successfully serialized and saved as 'california_housing_rf_model.pkl'")
```
